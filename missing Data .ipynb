```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna() # the places where the value is null there it will show True and other places it will
#it will show False.
```

```python
df.isna().sum() # sum will show which coulumn has how many null values.
```

```python
df.isna().any() # if a coloum does not have a null values it will show False else it will show True.
```
