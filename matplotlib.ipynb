```python
import pandas as pd
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#matplotlib  basic 
```

```python
x=np.linspace(0,5,11)
x
#y=x**2
y=x**3
y
```

```python
plt.plot(x,y)
plt.title("hello")
plt.xlabel("x")
plt.ylabel("y")
```

```python
#subplots in matplotlib
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
#object oriented way
```

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,.4,.8])
#how much move from left , bottom , then width , height.
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.2,.3,.3])
axis2.plot(x,y)
```

```python
fig=plt.figure(figsize=(20,8),dpi=500)
axis1=fig.add_axes([0.1,0.1,.4,.8])
#how much move from left , bottom , then width , height.
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.2,.3,.3])
axis2.plot(x,y)
```

```python
#types of plot
plt.scatter(x,y)
```

```python
plt.boxplot(x)
```

```python
#saving plots
```

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
